```python
pelamar = [("Budi", 3.2, 5), ("Siti", 3.6, 3), ("Dodi", 3.8, 1)]
for nama, ipk, semester in pelamar:
    if ipk >= 3.5 and semester > 2:
        status = "Diterima"
    else:
        status = "Ditolak"
    print(f"Nama: {nama} | Status: {status}")
```

```python
karyawan = [
    ("Joko", "IT", 7000),
    ("Rina", "HR", 5000),
    ("Toni", "Data", 8000),
    ("Lia", "IT", 6500),
]

total_gaji = 0

for nama, departemen, gaji in karyawan:
  # Cek apakah departemen adalah IT atau Data
  if departemen == "IT" or departemen == "Data":
    total_gaji += gaji
    print(f"[+] {nama} ({departemen}) : Rp{gaji:,}")

print("-" * 30)
print(f"Total Gaji Akhir : Rp{total_gaji:,}")
```

```python
penjualan = [
    ("Kopi", 60, False),
    ("Teh", 30, False),
    ("Boba", 80, True),
]

print("=== MINUMAN YANG MEMENUHI KRITERIA ===")

for nama_minuman, jumlah_terjual, sedang_promo in penjualan:
  # Kriteria: terjual > 50 DAN tidak sedang promo (not sedang_promo)
  if jumlah_terjual > 50 and not sedang_promo:
    print(f"[+] {nama_minuman} (Terjual: {jumlah_terjual} cup)")
```
